```python
X=data.iloc[:,3:13].values
```

```python
y=data.iloc[:,13].values
```

```python
from sklearn.preprocessing import LabelEncoder
labelencoder_X_1=LabelEncoder()
labelencoder_X_2=LabelEncoder()
```

```python
X[:,1]=labelencoder_X_1.fit_transform(X[:,1])
X[:,2]=labelencoder_X_1.fit_transform(X[:,2])
```

```python
#Dumy Encoding
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_transformer
```

```python
transformer=make_column_transformer((OneHotEncoder(drop='first'),[1,2]),
remainder='passthrough')
X=transformer.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=60)
```

```python
from sklearn.preprocessing import StandardScaler
sc_X=StandardScaler()
X_train=sc_X.fit_transform(X_train)
X_test=sc_X.fit_transform(X_test)
```

```python
#Yapay Sinir Ağı Modelini Oluşturmak
from keras.models import Sequential
from keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(units=4,kernel_initializer='uniform',activation='relu',input_dim=11)) #Girdi Katmanını Oluşturduk
model.add(Dense(units=4,kernel_initializer='uniform',activation='relu')) #Gizli Katmanı Oluşturduk. Her katmanda input_dim vermeye gerek yok
model.add(Dense(units=4,kernel_initializer='uniform',activation='relu'))
model.add(Dense(units=1,activation='sigmoid',kernel_initializer='uniform')) #Çıktı Katmanını Oluşturduk
model.summary()
```

```python
model.compile(optimizer='Adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
#Modelimizi Eğitelim
model.fit(X_train,y_train,epochs=10,batch_size=10)
```
